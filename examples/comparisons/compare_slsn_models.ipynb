```python
import matplotlib.pyplot as plt

from caat import DataCube, SN, SNModel
from caat.utils import WLE
from slsne.lcurve import get_all_lcs

slsn_model = SNModel(
    surface="SLSN-I_SLSN-I_GP_model.fits"
)
sn = SN(name="SN2020jhm")
datacube = DataCube(sn=sn)
datacube.construct_cube()
cube = datacube.cube

for band in ['B', 'g', 'r', 'i']:
    (low, mean, high), (time_samples, lightcurves) = get_all_lcs(band, return_individual=True, sigma=2)

    mean_lc_peak = min(mean)

    slsn_model.predict_lightcurve(phase_min=-50, phase_max=149, wavelength=WLE[band], show=False)
    filtered_cube = cube[
            (cube["ShiftedFilter"] == band)
            & (cube["Phase"] > -50)
            & (cube["Phase"] < 149)
            & (~cube["Nondetection"])
        ]

    observed_phases = filtered_cube["Phase"].values
    observed_fluxes = filtered_cube["ShiftedMag"].values
    observed_flux_errs = filtered_cube["ShiftedFluxerr"].values

    plt.errorbar(
        observed_phases,
        observed_fluxes,
        yerr=observed_flux_errs,
        color="lime",
        fmt="o",
        mec="k",
        label=sn.name,
    )    
    # Plot the mean range in r-band
    plt.fill_between(time_samples, mean_lc_peak-low, mean_lc_peak-high, color = 'red', alpha = 0.4, linewidth = 0, zorder = 2)
    plt.plot(time_samples, mean_lc_peak-mean, color = 'red', zorder = 3)
    plt.xlim(-50, 150)
    plt.ylim(-10, 2)
    # plt.gca().invert_yaxis()
    plt.show()
```

## Evaluation Framework

Compare GOPREAUX SLSN-I model against Gomez et al. 2024 on hold-out SNe not present in the Gomez training set.

```python
# Sample split: 48 SNe in both catalogs (training overlap), 6 unique to GOPREAUX (hold-out)
OVERLAP = [
    "SN2015bn", "SN2016aj", "SN2016ard", "SN2016els", "SN2017dwh", "SN2017egm",
    "SN2017gci", "SN2018bgv", "SN2018bym", "SN2018fcg", "SN2018gft", "SN2018hti",
    "SN2018ibb", "SN2018jfo", "SN2018kyt", "SN2018lfd", "SN2019cca", "SN2019enz",
    "SN2019eot", "SN2019gfm", "SN2019hge", "SN2019kcy", "SN2019lsq", "SN2019neq",
    "SN2019nhs", "SN2019pud", "SN2019sgg", "SN2019sgh", "SN2019szu", "SN2019unb",
    "SN2020abjc", "SN2020adkm", "SN2020aewh", "SN2020qlb", "SN2020rmv", "SN2020tcw",
    "SN2020uew", "SN2020xga", "SN2020zbf", "SN2020znr", "SN2021bnw", "SN2021ek",
    "SN2021hpx", "SN2021kty", "SN2021lwz", "SN2021zcl", "SN2022abdu", "SN2022lxd",
]

HOLD_OUT = [
    "SN2019hcc",
    "SN2019stc",
    "SN2020jhm",
    "SN2021ahpl",
    "SN2022acch",
    "SN2022xgc",
]

print(f"Overlap with Gomez et al. 2024: {len(OVERLAP)}")
print(f"Hold-out (GOPREAUX only):        {len(HOLD_OUT)}")
```

```python
import os
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
from scipy.interpolate import interp1d

from caat import SN


# Filters present in the GOPREAUX model and their Gomez band equivalents
# Extend/trim this list to match what the SLSN-I model was actually fit on
BAND_MAP = {
    # GOPREAUX filter : Gomez band name
    "g": "g",
    "r": "r",
    "i": "i",
    # "z": "z",
    "B": "B",
    "V": "V",
}

MIN_PHASE = -25
MAX_PHASE = 149
```

```python
from slsne.lcurve import get_all_lcs

# Pre-compute Gomez mean light curves for each band (full catalog = model prediction)
# get_all_lcs returns absolute magnitudes, so we normalize to delta-mag by
# subtracting the peak (minimum magnitude = maximum brightness) of the mean LC.
_gomez_cache = {}

def get_gomez_lc(band: str, min_time: float = -70, max_time: float = 300):
    """Return (time_samples, mag_mean, mag_lo, mag_hi) for a Gomez band,
    normalized to delta-magnitude (0 at peak), cached."""
    if band not in _gomez_cache:
        (lo, mean, hi), (times, _) = get_all_lcs(
            band,
            min_time=min_time,
            max_time=max_time,
            time_samples=500,
            sigma=1,
            return_individual=True,
        )
        peak = np.nanmin(mean)  # brightest point = minimum absolute magnitude
        _gomez_cache[band] = (times, mean - peak, lo - peak, hi - peak)
    return _gomez_cache[band]


def predict_gomez(obs: pd.DataFrame) -> pd.DataFrame:
    """
    Interpolate the Gomez catalog mean light curve at each observed phase point.
    Returns obs with added columns: Gomez_mag, Gomez_lo, Gomez_hi.
    Points outside the Gomez phase range are returned as NaN.
    """
    result = obs.copy()
    gomez_mags, gomez_lo, gomez_hi = [], [], []

    for _, row in obs.iterrows():
        band = row["CompFilter"]
        try:
            times, mean, lo, hi = get_gomez_lc(band)
            interp_mean = interp1d(times, mean, bounds_error=False, fill_value=np.nan)
            interp_lo   = interp1d(times, lo,   bounds_error=False, fill_value=np.nan)
            interp_hi   = interp1d(times, hi,   bounds_error=False, fill_value=np.nan)
            gomez_mags.append(float(interp_mean(row["Phase"])))
            gomez_lo.append(float(interp_lo(row["Phase"])))
            gomez_hi.append(float(interp_hi(row["Phase"])))
        except Exception:
            gomez_mags.append(np.nan)
            gomez_lo.append(np.nan)
            gomez_hi.append(np.nan)

    result["Gomez_mag"] = gomez_mags
    result["Gomez_lo"]  = gomez_lo
    result["Gomez_hi"]  = gomez_hi
    return result
```

```python
from util import compute_metrics, get_observed_points, load_datacube, predict_gopreaux

def evaluate_sn(sn_name: str, model, plot: bool = False) -> pd.DataFrame:
    """
    Run both model predictions for a single hold-out SN and return a
    tidy DataFrame of metrics per filter.

    Args:
        sn_name: IAU name (e.g. 'SN2019hcc')
        model:   loaded SNModel object

    Returns:
        DataFrame with columns:
            SN, Filter, n_points,
            GOPREAUX_bias, GOPREAUX_mad, GOPREAUX_iqr, GOPREAUX_chi2,
            Gomez_bias,    Gomez_mad,    Gomez_iqr,    Gomez_chi2
    """
    cube = load_datacube(sn_name, "SLSN-I", "SLSN-I", min_phase=MIN_PHASE, max_phase=MAX_PHASE)
    obs  = get_observed_points(cube, BAND_MAP)
    if obs.empty:
        print(f"  {sn_name}: no usable detections in BAND_MAP filters")
        return pd.DataFrame()

    obs = predict_gopreaux(model, obs, plot=plot)
    obs = predict_gomez(obs)

    rows = []
    for filt, group in obs.groupby("Filter"):
        gp_resid = (group["GOPREAUX_mag"] - group["ShiftedMag"]).values
        gz_resid = (group["Gomez_mag"]     - group["ShiftedMag"]).values
        errs     = group["Magerr"].values

        gp_metrics = compute_metrics(gp_resid, errs)
        gz_metrics = compute_metrics(gz_resid, errs)

        # Drop bands with fewer than 5 data points from the analysis
        if gp_metrics["n"] > 4:
            rows.append({
                "SN":              sn_name,
                "Filter":          filt,
                "n_points":        gp_metrics["n"],
                "GOPREAUX_bias":   gp_metrics["bias"],
                "GOPREAUX_mad":    gp_metrics["mad"],
                "GOPREAUX_iqr":    gp_metrics["iqr"],
                "GOPREAUX_chi2":   gp_metrics["chi2_reduced"],
                "Gomez_bias":      gz_metrics["bias"],
                "Gomez_mad":       gz_metrics["mad"],
                "Gomez_iqr":       gz_metrics["iqr"],
                "Gomez_chi2":      gz_metrics["chi2_reduced"],
            })

    return pd.DataFrame(rows)
```

```python
from caat import SN
from util import CUBE_ROOT

# Run the evaluation over all hold-out SNe
# Requires the SLSN-I model to be loaded as `slsn_model` (see cell above)

all_results = []
for sn in OVERLAP:
    sn_obj = SN(name=sn)
    cube_path = os.path.join(CUBE_ROOT, "SLSN-I", "SLSN-I", sn, f"{sn}_datacube_mangled.csv")
    if sn_obj.info.get("peak_mag", "") and os.path.exists(cube_path):
        print(f"Evaluating {sn}...")
        result = evaluate_sn(sn, slsn_model, plot=False)
        if not result.empty:
            all_results.append(result)

results = pd.concat(all_results, ignore_index=True)
results
```

```python
import matplotlib.pyplot as plt

# Summary: median MAD per model across all hold-out SNe and filters
summary = results.groupby("Filter")[["GOPREAUX_mad", "Gomez_mad", "GOPREAUX_chi2", "Gomez_chi2"]].median()
print("Median MAD and reduced chi^2 per filter (lower is better):\n")
print(summary.to_string())

# Bar chart: MAD by filter
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
x = range(len(summary))
width = 0.35

for ax, metric, label in zip(
    axes,
    [("GOPREAUX_mad", "Gomez_mad"), ("GOPREAUX_chi2", "Gomez_chi2")],
    ["Median MAD (mag)", "Median reduced χ²"],
):
    gp_vals = summary[metric[0]].values
    gz_vals = summary[metric[1]].values
    ax.bar([i - width/2 for i in x], gp_vals, width, label="GOPREAUX", color="steelblue")
    ax.bar([i + width/2 for i in x], gz_vals, width, label="Gomez et al.", color="tomato")
    ax.set_xticks(list(x))
    ax.set_xticklabels(summary.index)
    ax.set_ylabel(label)
    ax.set_xlabel("Filter")
    ax.legend()

fig.suptitle("GOPREAUX vs. Gomez et al. 2024 — Hold-out SLSN-I evaluation")
plt.tight_layout()
plt.show()
```
